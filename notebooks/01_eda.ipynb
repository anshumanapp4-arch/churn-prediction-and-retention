```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to path
sys.path.append(os.path.abspath('..'))
from generate_sample_data import generate_telco_churn_dataset
from src.data_cleaner import DataCleaner
```

## 1. Load Raw Data
We load the IBM Telco Customer Churn dataset.

```python
data_path = os.path.join('..', 'data', 'raw', 'telco_churn.csv')
if not os.path.exists(data_path):
    df_gen = generate_telco_churn_dataset()
    os.makedirs(os.path.dirname(data_path), exist_ok=True)
    df_gen.to_csv(data_path, index=False)

df_raw = pd.read_csv(data_path)
print("Raw Dataset Shape:", df_raw.shape)
df_raw.head()
```

## 2. Inspect Data Types and Missing Values

```python
df_raw.info()
print("\nTarget Churn Distribution:")
print(df_raw['Churn'].value_counts(normalize=True))
```

## 3. Data Cleaning
- Convert `TotalCharges` from string to numeric
- Handle missing values for `tenure == 0`
- Drop customer ID feature
- Encode `Churn` (Yes = 1, No = 0)

```python
cleaner = DataCleaner(target_col='Churn', id_col='customerID')
df_clean = cleaner.clean_data(df_raw)
processed_path = os.path.join('..', 'data', 'processed', 'processed_churn.csv')
cleaner.save_processed(df_clean, processed_path)
df_clean.head()
```

## 4. Exploratory Visualizations

```python
plt.figure(figsize=(8, 4))
sns.countplot(x='Contract', hue='Churn', data=df_clean, palette='Blues_d')
plt.title('Churn Count by Contract Type')
plt.show()

plt.figure(figsize=(8, 4))
sns.boxplot(x='Churn', y='tenure', data=df_clean, palette='Set2')
plt.title('Tenure vs Churn')
plt.show()
```
